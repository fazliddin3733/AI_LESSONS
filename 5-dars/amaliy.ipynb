```python
import pandas as pd
```

```python
df=pd.read_csv('titanic_data.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.nunique()
```

```python
df['age']=df['age'].fillna(df['age'].mean())
```

```python
df.isnull().sum()
```

```python
df['embarked']=df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df=df.drop(['deck'],axis=1)
```

```python
df.isnull().sum()
```

```python
df['embark_town'] = df['embark_town'].fillna('Southampton')
```

```python
df.isnull().sum()
```

```python
dummies=pd.get_dummies(df['sex'],prefix='ustun',dtype=int)
```

```python
dummies
```

```python
df=pd.concat([df.drop(columns=['sex']),dummies],axis=1)
```

```python
df.info()
```

```python
dummies=pd.get_dummies(df['embarked'],prefix='ustun',dtype=int)
```

```python
df=pd.concat([df.drop(columns=['embarked']),dummies],axis=1)
```

```python
df.info()
```

```python
dummies=pd.get_dummies(df['class'],prefix='ustun',dtype=int)
df=pd.concat([df.drop(columns='class'),dummies],axis=1)
```
